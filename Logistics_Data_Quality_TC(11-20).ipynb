### **`Test Case 11: Proof-of-Delivery Validation`**
---

#### **Objective :-**
`Check whether delivery records have a shipment ID, POD ID,
and a valid delivery attempt number.`

#### **Expected Result :-**
`Identify delivery records with missing POD information
or invalid shipment and attempt details.`

**Code cell 1: Check POD links**

```python
# Convert Attempt_No to numeric
df["Attempt_No_Num"] = pd.to_numeric(
    df["Attempt_No"], errors="coerce"
)

# Find records with a POD ID but missing shipment or attempt details
pod_issues = df[
    df["POD_ID"].notna() &
    (
        df["Shipment_ID"].isna() |
        df["Attempt_No_Num"].isna() |
        (df["Attempt_No_Num"] < 0)
    )
]

print("POD linking issues:", len(pod_issues))

pod_issues[
    ["Shipment_ID", "Tracking_ID", "Attempt_No", "POD_ID", "Status"]
].head(10)
```

**Code cell 2: Find delivery records without a POD ID**

```python
# Standardize status text
status = df["Status"].fillna("").astype(str).str.upper()

# Identify delivery-related events
delivery_rows = df[
    status.str.contains("DELIVER", na=False)
]

# Find delivery rows without POD information
missing_pod = delivery_rows[
    delivery_rows["POD_ID"].isna()
]

print("Delivery rows without POD ID:", len(missing_pod))

missing_pod[
    ["Shipment_ID", "Tracking_ID", "Attempt_No", "Status", "POD_ID"]
].head(10)
```

### **`Test Case 12: Normalize Exception and Failure Reason Codes`**
---

#### **Objective**
* Different labels may represent the same delivery problem.

* For example:
- cust_not_home
- Customer Unavailable
- Recipient Unavailable

* These labels should be standardized into one common value.

#### **Expected Result**
* Create a new column called Exception_Reason_Standard.

```python
# Convert exception reasons to lowercase and remove extra spaces
reason = (
    df["Exception_Reason"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

# Define standard names for common reasons
reason_map = {
    "cust_not_home": "CUSTOMER_UNAVAILABLE",
    "customer unavailable": "CUSTOMER_UNAVAILABLE",
    "recipient unavailable": "CUSTOMER_UNAVAILABLE",

    "bad_address": "ADDRESS_ISSUE",
    "address issue": "ADDRESS_ISSUE",
    "incomplete address": "ADDRESS_ISSUE",

    "crushed box": "DAMAGED_PARCEL",
    "damaged": "DAMAGED_PARCEL",
    "pkg_damaged": "DAMAGED_PARCEL",

    "weather": "WEATHER_DELAY",
    "heavy_rain": "WEATHER_DELAY",
    "cyclone alert": "WEATHER_DELAY"
}

# Apply standard names
df["Exception_Reason_Standard"] = reason.map(reason_map)

# Standardize other reasons that are not in the mapping
other_reason = (
    df["Exception_Reason"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
    .str.replace(r"[^A-Z0-9]+", "_", regex=True)
    .str.strip("_")
)

# Keep the standardized form of unmapped reasons
df.loc[
    df["Exception_Reason_Standard"].isna() & (reason != ""),
    "Exception_Reason_Standard"
] = other_reason

# Preserve missing values as missing
df.loc[
    reason == "",
    "Exception_Reason_Standard"
] = pd.NA

# Compare original and standardized values
df[
    ["Exception_Reason", "Exception_Reason_Standard"]
].drop_duplicates().head(20)
```

### **`Test Case 13: Compare Promised and Actual Delivery Time`**
---

#### **Objective**

* Compare Promised_Delivery_TS with the timestamp of the
actual delivery event.

### **Rules**

- Actual delivery time <= promised time: ON_TIME
- Actual delivery time > promised time: LATE
- Missing promised or actual time: UNKNOWN

### **Expected Result**

- Create a Delivery_Result column.

```python
# Convert timestamps into datetime format
# format="mixed" supports different timestamp formats in Pandas 2+
df["Event_TS_Parsed"] = pd.to_datetime(
    df["Event_TS"],
    errors="coerce",
    utc=True,
    format="mixed"
)

df["Promised_TS_Parsed"] = pd.to_datetime(
    df["Promised_Delivery_TS"],
    errors="coerce",
    utc=True,
    format="mixed"
)

# Standardize status text
status = (
    df["Status"]
    .fillna("")
    .astype(str)
    .str.upper()
    .str.replace(r"[^A-Z]", "", regex=True)
)

# Select delivery events
delivery_events = df[
    status.str.contains("DELIVER", na=False)
].copy()

# Set the initial result
delivery_events["Delivery_Result"] = "UNKNOWN"

# Check which rows have both timestamps
valid_times = (
    delivery_events["Event_TS_Parsed"].notna() &
    delivery_events["Promised_TS_Parsed"].notna()
)

# Mark records as on time when both timestamps are available
delivery_events.loc[
    valid_times,
    "Delivery_Result"
] = "ON_TIME"

# Mark records as late when actual delivery is after the promise
late = (
    valid_times &
    (
        delivery_events["Event_TS_Parsed"] >
        delivery_events["Promised_TS_Parsed"]
    )
)

delivery_events.loc[
    late,
    "Delivery_Result"
] = "LATE"

# Display the results
print(delivery_events["Delivery_Result"].value_counts())

delivery_events[
    [
        "Shipment_ID",
        "Tracking_ID",
        "Event_TS",
        "Promised_Delivery_TS",
        "Status",
        "Delivery_Result"
    ]
].head(10)
```

**Test Case 14: Identify Missing Pickup or Delivery Events**
--- 

#### **Objective**
- Identify records with missing event timestamps and shipments without a pickup-like or delivery-like event in the available data.

### **Expected Result**
- Missing events should be flagged as UNKNOWN / NOT FOUND,not automatically classified as failed deliveries.

**Code cell 1: Find missing timestamps**

```python
# Use the parsed timestamp column from Test Case 13
missing_event_time = df[
    df["Event_TS_Parsed"].isna()
]

print("Rows with missing event timestamps:", len(missing_event_time))

missing_event_time[
    ["Shipment_ID", "Tracking_ID", "Event_Code", "Status", "Event_TS"]
].head(10)
```

**Code cell 2: Check pickup and delivery events by shipment**

```python
# Standardize status text
status = df["Status"].fillna("").astype(str).str.upper()

# Identify pickup-like and delivery-like events
df["Has_Pickup"] = status.str.contains("PICK", na=False)
df["Has_Delivery"] = status.str.contains("DELIVER", na=False)

# Summarize events for each shipment
shipment_events = df.groupby("Shipment_ID").agg(
    Has_Pickup=("Has_Pickup", "any"),
    Has_Delivery=("Has_Delivery", "any")
).reset_index()

# Convert Boolean results into readable labels
shipment_events["Pickup_Check"] = shipment_events[
    "Has_Pickup"
].map({
    True: "FOUND",
    False: "UNKNOWN / NOT FOUND"
})

shipment_events["Delivery_Check"] = shipment_events[
    "Has_Delivery"
].map({
    True: "FOUND",
    False: "UNKNOWN / NOT FOUND"
})

shipment_events.head(10)
```

### **`Test Case 15: Detect Possible Event Timestamp Ordering Issues`**

---

#### **Objective**
- Find cases where the event timestamp is earlier than the previous event timestamp for the same shipment in the original dataset row order.

#### **Expected Result**
- Flag possible timestamp-order issues for investigation.

```python
# Create a copy of rows that have valid timestamps
events = df[
    df["Event_TS_Parsed"].notna()
].copy()

# Compare each event with the previous row for the same shipment
events["Previous_Event_TS"] = (
    events.groupby("Shipment_ID")["Event_TS_Parsed"].shift(1)
)

# Flag cases where the timestamp goes backwards
events["Time_Order_Issue"] = (
    events["Previous_Event_TS"].notna() &
    (
        events["Event_TS_Parsed"] <
        events["Previous_Event_TS"]
    )
)

# Keep rows with possible ordering issues
time_issues = events[
    events["Time_Order_Issue"]
]

print("Possible timestamp ordering issues:", len(time_issues))

time_issues[
    [
        "Shipment_ID",
        "Previous_Event_TS",
        "Event_TS",
        "Event_Code",
        "Status"
    ]
].head(10)
```

### **`Test Case 16: Exclude Test, Cancelled, and Voided Orders`**
---


#### **Objective**
- Exclude test, cancelled, and voided orders from standard business metrics while retaining the original dataset.

#### **Expected Result**
- Create a separate analysis dataset without these orders.

### **`Test Case 17: Validate Freight, Surcharge, COD, and Claim Amounts`**
---

#### **Objective**

1. Convert financial columns to numeric values.
2. Identify negative amounts.
3. Standardize currency values.

#### **Expected Result**
Financial records that need review should be identified.


**Code cell 1: Find negative amounts**

```python
# Financial columns in the dataset
amount_columns = [
    "Freight_Cost",
    "Surcharge",
    "COD_Amount",
    "Claim_Amount",
    "Declared_Value"
]

# Convert financial columns to numeric
for column in amount_columns:
    df[column + "_Num"] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

# Check whether any amount is negative
negative_amounts = df[
    (df[[column + "_Num" for column in amount_columns]] < 0)
    .any(axis=1)
]

print("Rows with negative amounts:", len(negative_amounts))

negative_amounts[
    ["Shipment_ID"] + amount_columns
].head(10)
```

**Code cell 2: Standardize currency**

```python
# Convert currency labels to uppercase
currency = (
    df["Currency"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

# Define standard currency names
currency_map = {
    "INR": "INR",
    "RS": "INR",
    "₹": "INR"
}

# Apply the mapping
df["Currency_Standard"] = currency.map(currency_map)

# Keep other non-empty values so they can be reviewed
df.loc[
    (currency != "") & df["Currency_Standard"].isna(),
    "Currency_Standard"
] = currency

print(df["Currency_Standard"].value_counts(dropna=False))
```

**Test Case 18: Link Return Shipments to Parent Shipments**
---

#### **Objective**

- Check whether Parent_Shipment_ID values exist in the Shipment_ID column.

#### **Expected Result**

- Identify records whose parent shipment cannot be found
in the current dataset.

```python
# Create a set of known shipment IDs
known_shipments = set(
    df["Shipment_ID"].dropna().astype(str).str.strip()
)

# Clean the parent shipment IDs
parent_ids = (
    df["Parent_Shipment_ID"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# Identify rows that have a parent ID
has_parent = parent_ids != ""

# Find parent IDs that are not present in the dataset
missing_parent_links = df[
    has_parent &
    ~parent_ids.isin(known_shipments)
]

print("Missing parent shipment links:", len(missing_parent_links))

missing_parent_links[
    ["Shipment_ID", "Parent_Shipment_ID", "Order_Status"]
].head(10)
```

### **`Test Case 19: Mask Customer Personal Information`**
---

#### **Objective**

- Mask customer names, phone numbers, and street addresses before sharing data in reports or public GitHub repositories.

#### **Expected Result**

- Create a separate masked dataset without exposing the original customer information.

**`Test Case 20: Duplicate Tracking IDs and Unresolved Locations`**
---


#### **Objective**
1. Identify tracking IDs connected to multiple active shipments.
2. Find missing or unresolved origin depots.
3. Find missing or unresolved destination zones.

#### **Expected Result**

- Produce lists of tracking IDs and location values that need further investigation.

**Code cell 1: Find tracking IDs linked to multiple active shipments**

```python
# Standardize order status
order_status = (
    df["Order_Status"]
    .fillna("")
    .astype(str)
    .str.upper()
    .str.strip()
)

# Select active shipment records
active_df = df[
    order_status == "ACTIVE"
].copy()

# Count different shipments associated with each tracking ID
tracking_check = (
    active_df.groupby("Tracking_ID")["Shipment_ID"]
    .nunique()
    .reset_index()
)

tracking_check.columns = [
    "Tracking_ID",
    "Shipment_Count"
]

# Identify tracking IDs linked to multiple shipments
duplicate_tracking = tracking_check[
    tracking_check["Shipment_Count"] > 1
]

print(
    "Tracking IDs linked to multiple active shipments:",
    len(duplicate_tracking)
)

duplicate_tracking.head(10)
```

**Code cell 2: Find unresolved origin and destination values**

```python
# Function to check whether a location needs review
def location_needs_review(value):

    if pd.isna(value):
        return True

    value = str(value).strip().upper()

    if value == "":
        return True

    if (
        "UNKNOWN" in value or
        "TBD" in value or
        "UNMAPPED" in value
    ):
        return True

    return False


# Find origin depot values that need review
bad_origin = df[
    df["Origin_Depot"].apply(location_needs_review)
]

# Find destination zone values that need review
bad_destination = df[
    df["Destination_Zone"].apply(location_needs_review)
]

print("Origin locations needing review:", len(bad_origin))
print("Destination locations needing review:", len(bad_destination))

print("\nOrigin depot values:")
print(bad_origin["Origin_Depot"].value_counts().head(10))

print("\nDestination zone values:")
print(bad_destination["Destination_Zone"].value_counts().head(10))
```
