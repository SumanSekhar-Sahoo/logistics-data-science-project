For testcase-10

```python
#Convert Timestamp Columns

df['Event_TS_UTC'] = pd.to_datetime(df['Event_TS_UTC'])

df['Promised_Delivery_TS'] = pd.to_datetime(df['Promised_Delivery_TS'])
```

```python
#Check Event Names

print(df['Event_Code'].unique())
```

```python
#Standardize Event_Code

df['Event_Code'] = (
    df['Event_Code']
      .str.upper()
      .str.strip()
      .str.replace(" ", "_")
)
```

```python
#Keep Only Delivered Shipments

delivered = df[
    df['Event_Code'].isin(
        ['DELIVERED', 'DELIV_SUCCESS']
    )
]
```

```python
#Compare Delivery Time with Promised Time

delivered['Delivery_Status'] = 'ON_TIME'

delivered.loc[
    delivered['Event_TS_UTC'] >
    delivered['Promised_Delivery_TS'],
    'Delivery_Status'
] = 'LATE'
```

```python
print(
    delivered[
        [
            'Tracking_ID',
            'Promised_Delivery_TS',
            'Event_TS_UTC',
            'Delivery_Status'
        ]
    ].head(20)
)
```

```python
print(delivered.columns)
```

```python
#Count ON_TIME and LATE Shipments

print(
    delivered['Delivery_Status']
    .value_counts()
)
```

```python
#Display Only Late Shipments

print(
    delivered[
        [
            'Tracking_ID',
            'Promised_Delivery_TS',
            'Event_TS_UTC',
            'Delivery_Status'
        ]
    ].head(10)
)
```

```python
delivered.to_csv(
    "TC10_Delivery_Promise.csv",
    index=False
)
```
